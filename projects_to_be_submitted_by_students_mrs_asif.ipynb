# **Project_02**
Guess the Number Game Python Project (computer)

```python
# project 2


import random
def guess(x):
    random_number = random.randint(1, x)
    guess = 0
    while guess != random_number:
      guess = int(input(f"Guess a number between 1 and {x}: "))
      if guess < random_number:
        print("Sorry, guess again. Too low.")
      elif guess > random_number:
        print("Sorry, guess again. Too high.")
    print(f"yay congrats you have guessed the number correctly {random_number}!")

guess(10)
```

# **Project_03**
** Guess the Number Game Python Project (user)**

```python

import random

def computer_guess(x):
    low = 1
    high = x
    feedback = ''

    while feedback != 'c':
        if low != high:
            guess = random.randint(low, high)
        else:
            guess = low

        feedback = input(f"Is the guess {guess} too High (H), too Low (L), or Correct (C)? ").lower()

        if feedback == 'h':
            high = guess - 1
        elif feedback == 'l':
            low = guess + 1

    print(f"Yay! The computer has guessed your number {guess} correctly!")

computer_guess(1000)
```

# **Project_04**
## Rock, paper, scissors Python Project

```python

import random

def play():
    user = input("What's your choice? ('r' for Rock, 'p' for Paper, 's' for Scissors): ").lower()

    while user not in ['r', 'p', 's']:
        print("Invalid choice! Please enter 'r', 'p', or 's'.")
        user = input("Try again: ('r' for Rock, 'p' for Paper, 's' for Scissors): ").lower()

    computer = random.choice(['r', 'p', 's'])
    print(f"Computer chose: {computer}")

    if user == computer:
        return "It's a tie!"
    if is_win(user, computer):
        return "You won!"
    return "You lost!"

def is_win(player, opponent):
    """Return True if the player beats the opponent"""
    return (player == 'r' and opponent == 's') or \
           (player == 's' and opponent == 'p') or \
           (player == 'p' and opponent == 'r')

# Loop to allow the user to play multiple rounds
while True:
    print(play())
    play_again = input("Do you want to play again? (y/n): ").lower()
    if play_again != 'y':
        print("Thanks for playing! Goodbye!")
        break
```

# **Project_05**
Hangman Python Project

```python

import random
import string

words =[
    "apple", "chair", "mountain", "river", "teacher", "computer", "library", "elephant", "bridge", "garden",
    "bicycle", "window", "candle", "notebook", "mirror", "moon", "pillow", "shadow", "ladder", "compass",
    "theater", "necklace", "thunder", "rainbow", "suitcase", "ocean", "island", "whisper", "treasure", "guitar","trumpet", "fjord", "meerkat", "chimera",
    "cornfield", "wisteria", "clover", "dome", "pagoda", "galleon", "chateau", "tundra", "parchment", "crystal",
    "geyser", "sloop", "acorn", "bramble", "cathedral", "fir", "jade", "lichen", "moonstone", "plume", "quartz",
    "redwood", "silhouette", "topiary", "vortex", "yew", "zeppelin", "antique", "bazaar", "carousel", "diorama",
    "epiphany", "flourish", "gondola", "horizon", "infusion", "jubilee", "keystone", "lattice", "mosaic", "nectar",
    "obsidian", "panorama", "quiver", "renaissance", "serenade", "traverse", "utopia", "vestige", "whimsy", "zephyr"]

def get_valid_word(words):
    word = random.choice(words).lower()  # Convert to lowercase
    while '-' in word or ' ' in word:
        word = random.choice(words).lower()
    return word

def hangman():
    word = get_valid_word(words)
    word_letters = set(word)  # Unique letters in the word
    alphabet = set(string.ascii_lowercase)  # Use lowercase for consistency
    used_letters = set()
    lives = 6

    print("\nWelcome to Hangman!")

    while len(word_letters) > 0 and lives > 0:
        print("\nYou have", lives, "lives left. Used letters:", ' '.join(used_letters))

        # Display guessed letters with _ for missing ones
        word_list = [letter if letter in used_letters else '_' for letter in word]
        print("Current word: ", ' '.join(word_list))

        user_letter = input("\nGuess a letter: ").lower()

        if user_letter in alphabet - used_letters:  # Valid new guess
            used_letters.add(user_letter)
            if user_letter in word_letters:
                word_letters.remove(user_letter)
            else:
                lives -= 1
                print("Incorrect guess!")
        elif user_letter in used_letters:
            print("You already guessed that letter.")
        else:
            print("Invalid character, please enter a valid letter.")

    # End of game messages
    if lives == 0:
        print("\n💀 You died! The word was:", word)
    else:
        print("\n🎉 Congratulations! You guessed the word:", word)

# Run the game
hangman()
```

# **Project_06**
Countdown Timer Python Project

```python
# Project 6: Countdown Timer
import time
import sys

def countdown(t):
    print("\nTimer is running...\n")  # Display message when the timer starts

    while t >= 0:  # Ensure 00:00 is shown
        mins, secs = divmod(t, 60)
        timer = f"\r{mins:02d}:{secs:02d}"  # Use \r to overwrite the same line
        sys.stdout.write(timer)
        sys.stdout.flush()  # Force update in some terminals
        time.sleep(1)
        t -= 1

    print("\nTimer completed!")  # Move to a new line after completion

try:
    t = int(input("Enter the time in seconds: "))
    if t < 0:
        print("Please enter a positive number.")
    else:
        countdown(t)
except ValueError:
    print("Invalid input! Please enter a valid integer.")

print("Thank you!")
```
